```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-2,3)
x = []

for i in n:
    if i < -1:
        x.append(-2)
    elif i <= 1:
        x.append(i)
    else:
        x.append(2)

plt.plot(n, x)
plt.grid(True)
plt.title("Signal 3")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-2,11)

x = []

for i in n:
    u1 = 1 if i >= 0 else 0
    u2 = 1 if i >= 3 else 0
    u3 = 1 if i >= 7 else 0

    x.append(u1 - u2 - 5*u3)

plt.stem(n, x)
plt.title("Signal 4")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-3,4)

x = []

for i in n:
    value = 0

    if i == 0:
        value += 1

    if i == 1:
        value += 3

    if i == -1:
        value += 5

    x.append(value)

plt.stem(n, x)
plt.title("Signal 5")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)
plt.show()
```
